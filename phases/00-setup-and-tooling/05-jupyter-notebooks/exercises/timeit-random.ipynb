```python
import numpy as np
import random

num_elements = 100000

print("Comparing list comprehension vs. NumPy for creating random number arrays:")

print("\nList comprehension:")
%timeit [random.random() for _ in range(num_elements)]

print("\nNumPy array:")
%timeit np.random.rand(num_elements)
```
